# Typecasting

Conversion of one data type into another data type is known as typecasting.

Type of Typecasting

1. Explicit Typecasting
2. Implicit Typecasting

Explicit typecasting :- The converison of one data type into another data type done via developer or manually is called explicit typecasting.

Implicit typecasting :- One data type converted into another by python interpreter automatically is know as Implicit typecasting.

```python
a = "1"
b = "2"
print(a+b)
```

```python
# Implicit Typecasting

a = "11"
b = "2"
print(int(a)+int(b))
```

```python
# Implicit Typecasting

a = "11"
b = 5
print(int(a)+b)
```

```python
# Explicit typecasting

a = 2.5
b = 1
print(a+b)
print(type(a+b))
```
