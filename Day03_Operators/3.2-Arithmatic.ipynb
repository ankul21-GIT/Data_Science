# Arithmatic Operator

Arithmetic operators are used with numeric values to perform common mathematical operations:

```python
a = 10
b = 5

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
print(a**b)
```

```python
num1 = int(input("Enter the first number : "))
num2 = int(input("Enter the second number : "))
print()
add = num1+num2
subtrac = num1-num2
mul = num1*num2
div = num1/num2
floordiv = num1%num2
modulus = num1//num2
exponen = num1**num2

print("The addition of num1 and num2 is : ", add)
print("The subtraction of num1 and num2 is : ", subtrac)
print("The Multiplication of num1 and num2 is : ", mul)
print("The Division of num1 and num2 is : ", div)
print("The floordiv of num1 and num2 is : ", floordiv)
print("The Modulus of num1 and num2 is : ", modulus)
print("The exponentiation of num1 and num2 is : ", exponen)
```
