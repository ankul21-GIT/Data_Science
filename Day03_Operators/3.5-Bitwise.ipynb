# Python Bitwise Operators

Bitwise operators are used to compare (binary) numbers

Bitwise operators are used to compare (binary) numbers

![![image.png](https://)](attachment:image.png)

```python
# Operator	Name	         Description	                               Example

# & 	     AND	Sets each bit to 1 if both bits are 1	                 x & y	
# |	         OR	    Sets each bit to 1 if one of two bits is 1	             x | y	
# ^	         XOR	Sets each bit to 1 if only one of two bits is 1	         x ^ y	
# ~	         NOT	Inverts all the bits	                                  ~x	
# <<	Zero fill left shift	     Shift left by pushing zeros in from the right and let the 
#                                     leftmost bits fall of                   x<<2
# >>	Signed right shift	        Shift right by pushing copies of the leftmost bit in from the
#                                   left, and let the rightmost bits fall of.  x>>2 
```

AND Operator

Sets each bit to 1 if both bits are 1

```python
a = 6
b = 3
print(a&b)
```

```python
print(7&5)
```

OR Operator

Sets each bit to 1 if one of two bits is 1

```python
a = 6
b = 3
print(a|b)
```

```python
print(8|9)
```

XOR Operator

Sets each bit to 1 if only one of two bits is 1

```python
a = 7
b = 4
print(a^b)
```

```python
a = 10
b = 12
print(a^b)
```

NOT Operator

Inverts all the bits

```python
a = ~6
print(a)
```

Zero fill left shift Operator

Shift left by pushing zeros in from the right and let the leftmost bits fall off

```python
a = 5
b = 3
print(a<<b)
```

```python
print(6<<2)
```

Signed right shift Operator

Shift right by pushing copies of the leftmost bit in from the left, and let the rightmost bits fall off

```python
a = 5
b = 1
print(a>>b)
```

```python
a = 5
b = 2
print(a>>b)
```

```python
a = 5
b = 3
print(a>>b)
```

```python
a = 9
b = 1
print(a>>b)
```
