# Slicing

You can return a range of characters by using the slice syntax.

```python
var = "Hello World"
print(var)
print(var[2:5])
print(var[1:8])
print(var[0:7])
```

Slice from the start = [:3]

```python
var = "Hello World"
print(var)
print(var[:4])
print(var[:5])
print(var[:8])
print(var[:])
```

Slice from end

```python
var = "Hello World"
print(var)
print(var[3:])
print(var[5:])
print(var[1:])
print(var[6:])
```

Negative indexing

```python
txt = [1,2,3,4,5,6,7]
print(txt)
print(txt[-6:-2])
print(txt[-4:-1])
print(txt[-4:])
```
