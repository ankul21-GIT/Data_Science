# String Cocatenation

```python
a = "Hello"
b = "World"
c = a+b
print(c)
```

```python
a = "Hello"
b = "World"
c = a+" "+b
print(c)
```

# Escape Sequence Character

```python
txt = "We are the so-called \"Vikings\" from the north."
print(txt)
```

```python
txt = "We are the so-called viking\nfrom the north."
print(txt)
```
