# Methods in String

- upper() :-	Converts a string into upper case
- lower() :-	Converts a string into lower case

```python
a = "Harry"
print(a)
print(len(a))
print()
print(a.upper())
print(a.lower())
```

- strip("") :- The strip() method removes any leading, and trailing whitespaces.

```python
name = "@Ankur!!"
print(name.strip("!,@"))
```

```python
txt = ",,,,,rrttgg.....banana....rrr"

x = txt.strip(",.grt")

print(x)
```

```python
txt = "   Ankur  "
print(txt.strip( ))
```

- rstrip("") :- The rstrip() method removes any trailing characters (characters at the end a string), space is the default trailing character to remove.

```python
txt = "!!Ankur!!!!!!"
print(txt.rstrip("!"))
```

```python
txt = "banana,,,,,ssqqqww....."
print(txt.rstrip(",.qsw"))
```

```python
txt = "     banana     "
x = txt.rstrip()
print("of all fruits", x, "is my favorite")
```

- lstrip("") :- The lstrip() method removes any leading characters (space is the default leading character to remove)

```python
txt = "!!Ankur!!!!!!"
print(txt.lstrip("!"))
```

```python
txt = "banana,,,,,ssqqqww....."
print(txt.lstrip(",.qsw"))
```

```python
txt = "     banana     "
x = txt.lstrip()
print("of all fruits", x, "is my favorite")
```

count() :- count method return the number of times the given value has occured within 
the given string.


```python
counts = "Hello sir feel good"
print(counts.count("l"))
```

```python
fruit = "I love apple, apple is my favourate fruit"
print(fruit.count("apple"))
```

zfill() :- The zfill() method adds zeros (0) at the beginning of the string, until it reaches the specified length.

If the value of the len parameter is less than the length of the string, no filling is done.

```python
z = "hello"
print(z.zfill(10))
```

```python
z = "hello"
print(z.zfill(4))
```

```python
a = "hello"
b = "welcome to the jungle"
c = "10.000"

print(a.zfill(10))
print(b.zfill(10))
print(c.zfill(10))
```

```python
txt = "50"

x = txt.zfill(10)

print(x)
```
