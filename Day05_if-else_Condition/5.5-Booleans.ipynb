# Boolean

Booleans represent one of two values: True or False.

```python
print(10 > 9)
print(10 == 9)
print(10 < 9)
```

```python
a = 200
b = 33

if b > a:
  print("b is greater than a")
else:
  print("b is not greater than a")
```

The bool() function allows you to evaluate any value, and give you True or False in return,

```python
print(bool("Hello"))
print(bool(15))
```

```python
x = "Hello"
y = 15

print(bool(x))
print(bool(y))
```

```python
def myFunction() :
  return True

if myFunction():
  print("YES!")
else:
  print("NO!")
```

```python
print(bool(False))
print(bool(None))
print(bool(0))
print(bool(""))
print(bool(()))
print(bool([]))
print(bool({}))
```

```python
print(bool(True))
print(bool("Ankur"))
print(bool(13))
print(bool("s"))
print(bool((1,2)))
print(bool([1,"Abhi",0]))
print(bool({1,2,3}))
```
