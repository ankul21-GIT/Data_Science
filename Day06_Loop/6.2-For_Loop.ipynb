# For loop

A for loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
```

```python
for x in "banana":
    print(x)
```

```python
for i in range(5):
    print(i)
```

```python
name = "Ankur"
for x in name:
    print(x)
```

```python
name = "Ankur"
for x in name:
    print(x)
    if x == "k":
        print("Something is special")
```

```python
for i in range(0,10):
    print(i)
```

```python
for i in range(0,20,2):
    print(i)
```

```python
for i in range(0,20,3):
    print(i)
```

```python
for i in range(10,0,-2):
    print(i)
```

```python
for i in range(10,1,-2):
    print(i)
```

```python
color = ["Red", "Green", "Blue", "Yellow"]
for i in color:
    print(i)
    for x in i:
        print(x)
```

```python
for i in range(5):
    print(i+1)
```

```python
string = "Ankur"

for i in range(3):
    for z in string:
        print(z)
```

```python
name = "Ankur"
for i in range(4):
    print(name)
```

```python
name = "Ankur Singh"
for i in name:
    print(i)
```

```python
for num in range(2, 101): 
    count = 0
    for i in range(1, num + 1):
        if num % i == 0:
            count += 1
    if count == 2:
        print(num)
```

```python
# Write a table

n = int(input("Enter any number : "))

print(f"The multiplication of the {n} is : ")

for i in range(1, 11):
    print(f"{n} * {i} = {n*i}")
    
```
