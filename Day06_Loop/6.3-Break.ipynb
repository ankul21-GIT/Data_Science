# Break statement

With the break statement we can stop the loop before it has looped through all the items

```python
for i in range(10):
    print(i)
```

```python
for i in range(10):
    print(i)
    if i == 5:
        break
```

```python
fruits = ["apple", "banana", "cherry", "Mango"]
for x in fruits:
  print(x)
  if x == "banana":
    break
```

```python
for i in range(20) :
    if i%2 == 14:
        break
    print(i)
```
