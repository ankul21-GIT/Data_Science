# Continue Statement

With the continue statement we can stop the current iteration of the loop, and continue with the next:

```python
fruits = ["Apple","Banana","Cherry","Mango"]
for x in fruits:
    if x == "Banana":
        continue
    print(x)
    
```

```python
for i in range(10):
    if i == 4:
        continue
    print(i)
```
