```python
for i in range(5):
    for j in range(5):
        print("*",end=" ")
        
    print()
```

```python
# left triangle
n = 5
for i in range(1, n+1):
    for k in range(1, i+1):
        print("*",end=" ")
    print()
```

```python
for i in range(5):
    for j in range(i):
        print("*",end=" ")
        
    print()
```

```python
n = 5  # Number of rows

for i in range(n):
    for j in range(i + 1):
        print("*", end=" ")
    print()
```

```python
for i in range(1, 5):
    print("*" * i)
```

```python
n = 5  # Number of rows

for i in range(n):
    # Print spaces
    for j in range(n - i - 1):
        print(" ", end=" ")
    # Print stars
    for k in range(i + 1):
        print("*", end=" ")
    print()
```

```python
n = 5  # Number of rows

for i in range(1, n + 1):
    spaces = "  " * (n - i)     # Two spaces for better alignment
    stars = "* " * i            # Star and space
    print(spaces + stars)
```

```python
# Pyramid

n = 5  # Number of rows

for i in range(n):
    # Print spaces
    for j in range(n - i - 1):
        print(" ", end=" ")
    # Print stars
    for k in range(2 * i + 1):
        print("*", end=" ")
    print()

```
