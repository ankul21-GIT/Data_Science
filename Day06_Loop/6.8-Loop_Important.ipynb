Find sum of first n natural number

```python
n = int(input("Enter any number : "))
sum = 0

for i in range(1, n+1):
    sum += i
print(f"The sum of the the {n} is : {sum}")
```

```python
def findSum(n):
    sum = 0
    x = 1
    
    # Iterating over all the numbers between 1 to n
    while x <= n:
        sum = sum + x
        x = x + 1
    return sum

if __name__ == "__main__":
    n = 6
    print(findSum(n))
```

```python
def findSum(n):

    # Using mathematical formula to compute
    return n * (n + 1) // 2

if __name__ == "__main__":
    n = int(input("Enter any number : "))
    print(findSum(n))
```

Factorial Number

```python
# factorial number

n = int(input("Enter any number : "))

total = 1

for i in range(1, n+1):
    total *= i

print(f"The factorial of the {n} is : {total} ")
```

```python
## Using Recursion
def fact(n):
    return 1 if (n==1 or n==0) else n * fact(n - 1) 

# Driver Code
num = 5
print(fact(num))
```

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n * factorial(n-1)
    
print(factorial(6))
```
