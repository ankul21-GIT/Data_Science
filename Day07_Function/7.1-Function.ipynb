# Function

- A function is a block of code that perform specific task whenever it is called. In bigger program, where we have large amount of code, it is advisable to create or use existing function that make the program flow organized and neat.

- A function is a block of code that performs a specific task. Functions help in organizing code, reusing code, and improving readability.

- A parameter is the variable listed inside the parentheses in the function definition.

- An argument is the value that is sent to the function when it is called.

A function is a block of code which only runs when it is called.

You can pass data, known as parameters, into a function.

A function can return data as a result.

# Type of function

- Built-in function
- User-define function

1. Built-in function

These function are defined and pre-coded in python. Some example of built-in function are:

min(), max(), sum(), range(), dict(), list(), len()

2. User-define function

We can create a function to perform specific task as per our need. Such function are called user-defined funtion

```python
# function with name add and parameters-(a,b,c)
def add(a,b,c):
    print(a+b+c)
    
# calling the same add function with arguments-(10,15,25)
add(10,15,25)
```

```python
## why functions?
num=24
if num%2==0:
    print("the number is even")
else:
    print("the number is odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("the number is even")
    else:
        print("the number is odd")
        
even_or_odd(5)
```

```python
def calculator(a, b):
    print("addition : ", a + b)
    print("Subtraction : ", a-b)
    print("Division : ", a/b)
    print("Multiplication : ", a * b)
    
calculator(8, 4)
print()
calculator(10, 4)
print()
calculator(8, 2)
```

```python
calculator(3,2)
print()
calculator(6,8)
print()
calculator(8, 4)
```
