# Lambda Function

Lambda functions are small anonymous functions defined using the lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.

```python
def double(x):
    return x*2

print(double(5))
```

```python
double = lambda x: x*2
print(double(5))
```

```python
double = lambda x: x*2
cube = lambda x: x*x*x

print(double(5))
print(cube(5)) 
```

```python
double = lambda x: x*2
cube = lambda x: x*x*x
avg = lambda x,y,z: (x+y+z)/2

print(double(5))
print(cube(5)) 
print(avg(3, 5, 10))
```

```python
def appl(fx, value):
    return 6 + fx(value)

cube = lambda x: x*x*x
print(appl(cube, 2))
```

```python
def appl(fx, value):
    return 6 + fx(value)
print(appl(lambda x: x * x, 2))
```

```python
addition=lambda a,b:a+b
type(addition)
print(addition(5,6))
```

```python
def even(num):
    if num%2==0:
        return True
    else:
        return False
    
even(24)
```
