# Union()

- union() :- In the union method same element not print.

```python
s1 = {1, 2, 5, 6, 7}
s2 = {3, 6, 7, 4, 1}
print(s1.union(s2))
```

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.union(y)

print(z)
```

# Intersect()

- Intersect() :- It print only that element which are same in both variable s1 and s2




```python
s1 = {1, 2, 5, 6, 7}
s2 = {3, 6, 7, 4, 1}
print(s1.intersection(s2))
```

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.intersection(y)

print(z)
```

# update()

- The update() method updates the current set, by adding items from another set (or any other iterable).

```python
s1 = {1, 2, 5, 6, 7}
s2 = {3, 6, 7, 4, 1}
s1.update(s2)
print(s1, s2)
```

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

x.update(y)

print(x)
```

# symmetric_difference

- The returned set contains a mix of items that are not present in both sets.

```python
# symmetric_difference() :- Not print the common value.

cities1 = {"Tokyo", "Madrid", "Berlin", "Delhi"}
cities2 = {"Tokyo", "Seoul", "Kabul", "Madrid"}
cities3 = cities1.symmetric_difference(cities2)
print(cities3)
```

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.symmetric_difference(y)

print(z)
```

# difference()

- The difference() method returns a set that contains the difference between two sets.

```python
# difference() :- The value which are difference in both.

cities1 = {"Tokyo", "Madrid", "Berlin", "Delhi"}
cities2 = {"Seoul", "Kabul", "Delhi"}
cities3 = cities1.difference(cities2)
print(cities3)
```

```python
cities1 = {"Tokyo", "Madrid", "Berlin", "Delhi"}
cities2 = {"Seoul", "Kabul", "Delhi"}
cities3 = cities2.difference(cities1)
print(cities3)
```

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.difference(y)

print(z)
```

# isdisjoint()

- The isdisjoint() method returns True if none of the items are present in both sets, otherwise it returns False.

```python
# isdisjoint() :- when both the set are differance this is called disjoint.

cities1 = {"Tokyo", "Madrid", "Berlin", "Delhi"}
cities2 = {"Seoul", "Kabul", "Delhi1", "Benglore"}
print(cities1.isdisjoint(cities2))
```

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "facebook"}

z = x.isdisjoint(y)

print(z)
```

# issuperset()

- The issuperset() method returns True if all items in the specified set exists in the original set, otherwise it returns False.

```python
# issuperset() :- The issuperset() method check if all the item of a participant set are present in the orignal set. It return true if all the item are present, else it return false.

cities1 = {"Tokyo", "Madrid", "Berlin", "Delhi"}
cities2 = {"Tokyo"}
print(cities1.issuperset(cities2))
cities3 = {"Tokyo", "Madrid", "Delhi"}
print(cities1.issuperset(cities3))
print(cities3.issubset(cities1))
```

# issubset()

- The issubset() method returns True if all items in the set exists in the specified set, otherwise it returns False.

```python
x = {"a", "b", "c"}
y = {"f", "e", "d", "c", "b", "a"}

z = x.issubset(y)

print(z)
```

# remove()

- The remove() method removes the specified element from the set.

```python
cities = {"Tokyo", "Madrid", "Berlin", "Delhi"}
cities.remove("Tokyo")
print(cities)
```

```python
fruits = {"apple", "banana", "cherry"}

fruits.remove("banana")

print(fruits)
```

# discard()

- The discard() method removes the specified item from the set.

```python
cities = {"Tokyo", "Madrid", "Berlin", "Delhi"}
cities.discard("Tokyo2")
print(cities)
```

```python
fruits = {"apple", "banana", "cherry"}

fruits.discard("banana")

print(fruits)
```

# pop()

- The pop() method removes a random item from the set.

```python
# pop() :- These element remove the last element of the set.

cities = {"Tokyo", "Madrid", "Berlin", "Delhi"}
cities.pop()
print(cities)
```

```python
fruits = {"apple", "banana", "cherry"}

fruits.pop()

print(fruits)
```

```python
# del :- del is not a method, rather it is  a keyword which deletes the set entirely.

cities = {"Tokyo", "Madrid", "Berlin", "Delhi"}
del cities
print(cities)
```
