```python
student={"name":"Rishi","age":22,"grade":'A'}
student_copy=student
print(student)
print(student_copy)
```

```python
student={"name":"Rishi","age":22,"grade":'A'}
student["name"]="Krishanu"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() ## deep copy
print(student_copy1)
print(student)
```

```python
student["name"]="KRish3"
print(student_copy1)
print(student)
```
