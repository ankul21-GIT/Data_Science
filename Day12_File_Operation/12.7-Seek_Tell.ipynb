```python
## seek
with open("file11.txt", "r") as f:
    f.seek(10)
    
    data = f.read(12)
    print(data)
```

```python
## tell

with open("file11.txt", "r") as f:
    f.seek(10)
    
    print(f.tell())
    data = f.read(12)
    print(data)
```

```python
## truncate

with open('sample888.txt', 'w') as f:
    f.write('Hello World')
    f.truncate(5)
    
with open('sample888.txt', 'r') as f:
    print(f.read())
```
