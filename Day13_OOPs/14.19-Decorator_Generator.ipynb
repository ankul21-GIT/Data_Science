# Decorator

Decorator is function that take another function as argument and return a funtion.

```python
def hello():
    print("hello world")

hello()
```

```python
def greet(fx):
    def mfx():
        print("good morning")
        fx()
        print("Thanks for using these function")
    return mfx

@greet
def hello():
    print("Hello World")

hello()        
    
```

```python
def decorator(func):
    def wrapper():
        print("Transaction is initiated")
        func()
        print("Transaction is completed")
    return wrapper

def hello():
    print("Money is deducted from your account")

hello1 = decorator(hello)
hello()
    
```

```python
def decorator(func):
    def wrapper():
        print("Transaction is initiated")
        func()
        print("Transaction is completed")
    return wrapper

@decorator
def hello():
    print("Money is deducted from your account")

hello()
```

```python
def greet(fx):
    def mfx(*args, **kwargs):
        print("good morning")
        fx(*args, **kwargs)
        print("Thanks for using these function")
    return mfx

@greet
def hello():
    print("Hello World")

def add(a, b):
    print(a+b)

hello()        
greet(add)(2, 2)
    
```

## Generator

In Python, a generator is a special type of function or expression that creates an iterator, allowing for efficient, on-demand generation of values in a sequence. Unlike regular functions that return a single value and terminate, generators "yield" values one at a time, pausing execution and maintaining their state until the next value is requested.

```python
def mult():
    list= []
    i = 1
    while i<=200:
        list.append(i)
        i +=1
    return list

print(mult())
```

```python
import sys
def mult():
    list= []
    i = 1
    while i<=200:
        list.append(i)
        i +=1
    return list

print(mult())

z = sys.getsizeof(list)
print(z)
```

```python
import sys
def mult():
    list= []
    i = 1
    while i<=200:
        list.append(i)
        i +=1
    return list

print(mult())
z = sys.getsizeof(list)
print(z)

print([num+10 for num in mult()])
```

```python
def fun():
    yield 1            
    yield 2            
    yield 3
    yield 4
# Driver code to check above generator function
for val in fun(): 
    print(val)
```

```python
def fun(max):
    cnt = 1
    while cnt <= max:
        yield cnt
        cnt += 1

ctr = fun(9)
for n in ctr:
    print(n)
```
