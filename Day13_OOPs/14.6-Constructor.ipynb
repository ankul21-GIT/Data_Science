## Constructor
The constructor in Python is defined using the special method name __init__. 

```python
class Person:
    name = "Ankul"
    occupation = "Data Scientist"
    
    def info(self):
        print(f"{self.name} is a {self.occupation}")
        
a = Person()
a.info()
```

```python
class Person:
    name = "Ankul"
    occupation = "Data Scientist"
    
    def info(self):
        print(f"{self.name} is a {self.occupation}")
        
a = Person()
# Change the name
a.name = "Srishty"
a.occupation = "HR"
a.info()
```

```python
class Person:
    def __init__(self):
        print("Hey I am a person")
          
a = Person()
```

```python
class Person:
    def __init__(self, name, occ):
        print("Hey I am a person")
        self.name = name
        self.occ = occ
        
    def info(self):
        print(f"{self.name} is a {self.occ}")
          
a = Person("Ankul", "Data Scientist")
b = Person("Ram JI", "HR")
a.info()
b.info()
```
