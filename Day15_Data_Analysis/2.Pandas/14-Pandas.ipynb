## Pivot Table and Melt function

Melt

```python
import pandas as pd
```

```python
var = pd.DataFrame({"day":[1,2,3,4,5,6],"eng":[10,12,14,15,16,12],"math":[17,18,19,13,14,16]})
var
```

```python
pd.melt(var)
```

```python
pd.melt(var,id_vars=['eng'])
```

```python
pd.melt(var,id_vars=['day'])
```

```python
pd.melt(var,id_vars=['day'],var_name="python")
```

```python
pd.melt(var,id_vars=['day'],var_name="python",value_name="wscube")
```

Pivot

```python
var1 = pd.DataFrame({"day":[1,2,3,4,5,6],
                    "st_name":['a','b','c','a','b','c'],"eng":[10,12,14,15,16,12],"math":[17,18,19,13,14,16]})

var1
```

```python
var1.pivot(index='day',columns='st_name')
```

```python
var1 = pd.DataFrame({"day":[1,2,3,4,5,6],
                    "st_name":['a','b','a','a','b','b'],"eng":[10,12,14,15,16,12],"math":[17,18,19,13,14,16]})

var1
```

```python
var1.pivot(index='day',columns='st_name')
```

```python
## we only wanna see the data of the english
## we can access the data of the particulat columns

var1.pivot(index='day',columns='st_name', values='eng')
```

```python
var1 = pd.DataFrame({"day":[1,1,1,1,2,2],
                    "st_name":['a','b','a','a','b','b'],"eng":[10,12,14,15,16,12],"math":[17,18,19,13,14,16]})

var1
```

```python
var1.pivot_table(index='st_name',columns='day')
```

```python
## we can use the aggfunc to perform the mean max minimu 
var1.pivot_table(index='st_name',columns='day',aggfunc="max")
```

```python
## margin give the average value
var1.pivot_table(index='st_name',columns='day',aggfunc="sum",margins="True")
```

```python
## margin give the average value
var1.pivot_table(index='st_name',columns='day',aggfunc="mean",margins="True")
```
