# Mean

The mean is the average of a set of numbers. It is calculated by adding all the values in the dataset and dividing the sum by the total number of values.

```python
import numpy as np
data = [10, 20, 30, 40, 50]

mean = np.mean(data)

print("Mean:", mean)
```

***Mean From Pandas***

```python
import pandas as pd

data = {
    "Marks": [70, 80, 90, 60, 75]
}

df = pd.DataFrame(data)

print(df["Marks"].mean())
```

```python
# Visualization of Mean
import matplotlib.pyplot as plt

data = [10,20,30,40,50]

plt.plot(data, marker="o")

plt.axhline(np.mean(data))

plt.title("Mean Visualization")

plt.show()
```
