# Median

The median is the middle value in a dataset when the numbers are arranged in ascending or descending order. If the number of values is even, the median is the average of the two middle numbers.

```python
import numpy as np

data = [10, 20, 30, 40, 50]

median = np.median(data)

print("Median:", median)
```

***Pandas Median***

```python
import pandas as pd

data = {
    "Marks":[70,80,90,60,75]
}

df = pd.DataFrame(data)

print(df["Marks"].median())
```
