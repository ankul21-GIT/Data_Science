# Mode

The mode is the value that appears most frequently in a dataset. A dataset may have one mode, more than one mode, or no mode.

```python
from scipy import stats

data = [1,2,2,3,4,4,4,5]

mode = stats.mode(data)

print("Mode:", mode)
```

***Pandas Mode***

```python
import pandas as pd

data = {
    "Numbers":[1,2,2,3,4,4,4,5]
}

df = pd.DataFrame(data)

print(df["Numbers"].mode())
```
