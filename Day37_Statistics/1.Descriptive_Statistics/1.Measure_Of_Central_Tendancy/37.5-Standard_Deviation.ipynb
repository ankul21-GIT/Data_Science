# Standard Deviation

Variance gives spread, but the units become squared, which is hard to interpret.

So we use Standard Deviation, which is the square root of variance.

\sigma = \sqrt{\sigma^2}



```python
# Standard Deviation
import numpy as np

data = [10,12,14,16,18]

std = np.std(data)

print("Standard Deviation:", std)
```

```python
# Visualization of Standard Deviation
import matplotlib.pyplot as plt
import numpy as np

data = [10,12,14,16,18]

plt.hist(data)

plt.title("Data Distribution")

plt.show()
```
