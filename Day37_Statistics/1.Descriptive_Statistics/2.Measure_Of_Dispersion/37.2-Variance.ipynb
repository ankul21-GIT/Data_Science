# Variance
Variance measures how spread out the data is from the mean.
If variance is small → values are close to the mean.
If variance is large → values are spread far from the mean.

\sigma^2 = \frac{\sum (x_i - \mu)^2}{N}

Where

- 𝑥 = each data value
- μ = mean of the dataset
- N = number of observations

```python
# Variance using NumPy
import numpy as np

data = [10, 12, 14, 16, 18]

variance = np.var(data)

print("Variance:", variance)
```

```python
# Using Pandas
import pandas as pd

data = {"Numbers":[10,12,14,16,18]}

df = pd.DataFrame(data)

print(df["Numbers"].var())
```
