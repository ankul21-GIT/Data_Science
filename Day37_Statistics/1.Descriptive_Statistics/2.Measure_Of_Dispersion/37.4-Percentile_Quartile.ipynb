# Percentile

A percentile tells us the value below which a certain percentage of data falls.

```python
import numpy as np

data = [10,20,30,40,50]

p25 = np.percentile(data, 25)
p50 = np.percentile(data, 50)
p75 = np.percentile(data, 75)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
```

# Quartiles

Quartiles divide the dataset into 4 equal parts.

- Quartile	Meaning
- Q1	25% data
- Q2	50% data (Median)
- Q3	75% data

```python
import numpy as np

data = [10,20,30,40,50]

Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)

print("Q1:", Q1)
print("Median:", Q2)
print("Q3:", Q3)
```

# Interquartile Range (IQR)

IQR tells us the spread of the middle 50% of data.

Formula:

 - IQR=Q3−Q1

```python
import numpy as np

data = [10,20,30,40,50]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("IQR:", IQR)
```

# Outlier Detection

Outliers are extreme values in the dataset.

Rule:

- Lower limit

    Q1 - 1.5 * IQR

- Upper limit

    Q3 + 1.5 * IQR

```python
import numpy as np

data = [10,12,14,15,18,100]

Q1 = np.percentile(data,25)
Q3 = np.percentile(data,75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

print("Lower limit:", lower)
print("Upper limit:", upper)
```
