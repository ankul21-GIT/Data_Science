
# Skewness

Skewness tells us whether the data distribution is symmetric or tilted (skewed).

If the data is perfectly balanced around the mean → skewness = 0

There are three types of skewness.

1️⃣ Normal Distribution (No Skew)


Characteristics:

- Mean = Median = Mode

- Symmetrical distribution

- Skewness = 0

This distribution is very common in statistics and ML.

Examples:

- IQ scores

- Human height

- Measurement errors

2️⃣ Positive Skew (Right Skew)


Characteristics:

- Tail is on the right side

- Mean > Median > Mode

Examples:

- Income distribution

- House prices

- Company profits

A few very large values pull the mean to the right.

3️⃣ Negative Skew (Left Skew)


Characteristics:

- Tail is on the left side

- Mean < Median < Mode

Example:

- Difficult exam scores

- Retirement age

Most values are high but a few very small values pull the mean left.

```python
import pandas as pd

data = [10,12,13,14,15,100]

df = pd.Series(data)

print("Skewness:", df.skew())
```

# Kurtosis

Kurtosis measures how heavy the tails of a distribution are.

It tells us whether data has:

extreme outliers

heavy tails

flat distribution

Types of Kurtosis
1️⃣ Mesokurtic (Normal)

- Normal distribution.

Kurtosis ≈ 3

2️⃣ Leptokurtic

Very peaked distribution

- Heavy tails

- More outliers

Example:

Stock market returns

3️⃣ Platykurtic

- Flat distribution

- Light tails

Fewer outliers

```python
import pandas as pd

data = [10,12,13,14,15,100]

df = pd.Series(data)

print("Kurtosis:", df.kurt())
```
