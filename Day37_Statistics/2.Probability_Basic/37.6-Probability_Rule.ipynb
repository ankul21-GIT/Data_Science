# 1️. Addition Rule of Probability

The addition rule tells us the probability that either event A or event B occurs.


Where:

- P(A∪B) = probability of A or B

- P(A∩B) = probability of both A and B

Example

A card is drawn from a deck.

Find the probability of getting:

- a King

- or a Heart

Values:

- P(King)=4/52
- P(Heart)=13/52
- P(King∩Heart)=1/52

So,

P(A or B) = 4/52 + 13/52 − 1/52
          = 16/52

```python
P_A = 4/52
P_B = 13/52
P_AB = 1/52

result = P_A + P_B - P_AB

print(result)
```

# Multiplication Rule

The multiplication rule calculates the probability that both events happen together.



Example

Probability of:

    - Head in first toss = 1/2

    - Head in second toss = 1/2

Probability of Head AND Head:

1/2 × 1/2 = 1/4

```python
P_head = 1/2

P_both = P_head * P_head

print(P_both)
```

# Independent Events

Events are independent if one event does not affect the other.

Examples:

  - Coin toss

  - Dice roll

  - Lottery draw

Example:

Coin toss + dice roll

The result of the coin does not change the dice result.

# Dependent Events

Events are dependent if the first event affects the second.

Example:

Drawing cards without replacement.

Example:

Probability of drawing 2 Kings in a row

First draw:

4 / 52

Second draw:

3 / 51

So,

(4/52) × (3/51)

```python
P_first = 4/52
P_second = 3/51

P_two_kings = P_first * P_second

print(P_two_kings)
```
