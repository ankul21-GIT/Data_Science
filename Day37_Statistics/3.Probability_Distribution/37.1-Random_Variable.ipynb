# Random Variables

A Random Variable is a variable whose value depends on the outcome of a random experiment.

Example experiments:

Tossing a coin

Rolling a dice

Measuring rainfall

Counting number of customers

Instead of writing outcomes like Head/Tail, we assign numbers.



**Types of Random Variables**

There are two main types.

1. Discrete Random Variable

A variable that takes countable values.

Examples:

Number of students

Number of cars

Dice result

Number of emails received

Example:

Dice outcomes

1,2,3,4,5,6

These are countable numbers.

2. Continuous Random Variable

A variable that can take infinite decimal values.

Examples:

Height

Weight

Temperature

Time

Example:

170.2 cm
170.25 cm
170.256 cm

Values can continue infinitely.

2. Probability Mass Function (PMF)

PMF is used for discrete random variables.

It tells us the probability of each possible value.



```python
import numpy as np

dice = [1,2,3,4,5,6]

probabilities = [1/6]*6

for i,p in zip(dice, probabilities):
    print(i, p)
```

3. Probability Density Function (PDF)

PDF is used for continuous random variables.

Instead of probability at exact values, we calculate probability over intervals.

Example:

Probability that height is between:

170 cm and 175 cm

In continuous distributions:

P(X = exact value) = 0

Instead we calculate:

P(a < X < b)
Example of Continuous Distribution

Human height follows approximately normal distribution.



```python
from scipy.stats import norm

x = norm.pdf(0,0,1)

print(x)
```
