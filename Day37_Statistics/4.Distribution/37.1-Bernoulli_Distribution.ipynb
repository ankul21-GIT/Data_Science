# Bernoulli Distribution

 - A Bernoulli distribution is a discrete probability distribution for a random experiment that has only two possible outcomes:

Success (1)

Failure (0)

The probability of success is denoted by p, and the probability of failure is 1 − p.

Examples:

- Coin toss (Head = 1, Tail = 0)

- Email spam detection (Spam = 1, Not spam = 0)

- Pass or Fail in an exam

Formula of Bernoulli Distribution

P(X=x)=p^x(1-p)^{1-x}

Where:

- p = probability of success

- 1 − p = probability of failure

- x = outcome (0 or 1)



**Using NumPy**

```python
import numpy as np

# probability of success
p = 0.6

# generate Bernoulli random samples
data = np.random.binomial(1, p, 10)

print(data)
```

**Using SciPy**

```python
from scipy.stats import bernoulli

p = 0.6

samples = bernoulli.rvs(p, size=10)

print(samples)
```

**Visualization in Python**

```python
import matplotlib.pyplot as plt
from scipy.stats import bernoulli

p = 0.6

data = bernoulli.rvs(p, size=1000)

plt.hist(data, bins=2)
plt.title("Bernoulli Distribution")
plt.show()
```
