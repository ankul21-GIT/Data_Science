# P-value

The p-value is the probability of obtaining results at least as extreme as the observed data, assuming that the null hypothesis (H₀) is true.

In simple words:
The p-value tells us how strong the evidence is against the null hypothesis.



Example

Suppose a company claims:

Average battery life = 10 hours

We test this claim.

After performing a statistical test we get:

p-value = 0.02

Since:

0.02 < 0.05

We reject the null hypothesis.

Conclusion:
The battery life is significantly different from 10 hours.

# Intuition

Think of p-value like this:

- Small p-value → data is unlikely under H₀

- Large p-value → data is consistent with H₀

So:

Small p-value → strong evidence against the null hypothesis.

```python
from scipy import stats

data = [52,55,50,48,54,53,51,56,52,49]

t_stat, p_value = stats.ttest_1samp(data, 50)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```
