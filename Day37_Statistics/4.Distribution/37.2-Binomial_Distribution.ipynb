# Binomial Distribution

Binomial distribution is used when an experiment has only two possible outcomes.

Examples:

Coin toss → Head or Tail

Exam result → Pass or Fail

Email → Spam or Not Spam

Product → Defective or Non-defective

So each trial has two outcomes:

Success

Failure

Conditions of Binomial Distribution

A problem follows binomial distribution if:

1. Fixed number of trials (n)
2. Only two outcomes (success/failure)
3. Probability of success remains constant
4. Trials are independent

Example

Suppose:

A coin is tossed 5 times

We want probability of 3 heads

Here:

n = 5

p = 0.5

x = 3

Binomial Formula

P(X=k) = \binom{n}{k} p^k (1-p)^{n-k}

Where:

n = number of trials

k = number of successes

p = probability of success

(1-p) = probability of failure

Example Calculation

Coin toss:

n = 5

p = 0.5

k = 3

So:

P(X=3) = 10 × (0.5)^3 × (0.5)^2

Result:

0.3125

```python
from scipy.stats import binom

n = 5
p = 0.5
k = 3

probability = binom.pmf(k, n, p)

print(probability)
```
