# Poisson Distribution

Poisson distribution is used to model how many times an event happens in a fixed interval of time or space.

Examples:

Number of calls in a call center per hour

Number of website visits per minute

Number of accidents at a traffic signal per day

Number of emails received in an hour

So Poisson answers this type of question:

👉 How many events occur in a fixed interval?

Poisson Distribution Formula

P(X=k) = \frac{e^{-\lambda} \lambda^k}{k!}

Where:

λ (lambda) = average number of events

k = number of occurrences

e = Euler’s constant (~2.718)

Example

Suppose:

Average number of calls per hour = 4

So:

λ = 4

Find probability that exactly 2 calls occur.

k = 2

Substitute in formula.

Result ≈ 0.1465

So probability of 2 calls in an hour = 0.1465.

```python
from scipy.stats import poisson

lam = 4
k = 2

probability = poisson.pmf(k, lam)

print(probability)
```

**When Do We Use Poisson Distribution?**

When events:

1. occur randomly
2. occur independently
3. happen within a fixed time or space interval
