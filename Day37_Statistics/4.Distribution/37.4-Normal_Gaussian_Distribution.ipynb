# Z-Score (Standardization)

Z-score tells how many standard deviations a value is away from the mean.



Example

Suppose:

Mean = 70

Std Dev = 10

Score = 85

Z-score:

(85 - 70) / 10 = 1.5

So the value is 1.5 standard deviations above the mean.



```python
import numpy as np
from scipy.stats import norm

mean = 0
std = 1

x = norm.pdf(0, mean, std)

print(x)
```

```python
# Python Z-Score example
from scipy.stats import zscore
import numpy as np

data = [60,65,70,75,80,85]

print(zscore(data))
```
