# Log Normal Distribution

Lognormal Distribution is a crucial probability distribution model as it characterises data with positive values that show right-skewed patterns, which makes it suitable for various real-world scenarios like stock prices, income, resource reserves, social media, etc. Understanding Lognormal Distribution helps in risk assessment, portfolio optimisation, and decision-making in fields, like finance, economics, and resource management.



A Log-Normal Distribution is a distribution where the logarithm of the variable follows a normal distribution.

In simple words:

👉 If ln(X) is normally distributed, then X follows a log-normal distribution.

Mathematical Definition

\ln(X) \sim N(\mu,\sigma^2)

This means:

X is log-normally distributed

ln(X) follows normal distribution

**Shape of Log-Normal Distribution**

Characteristics:

Right-skewed distribution

Values cannot be negative

Long right tail

This means:

X > 0

The variable is always positive.





```python
import numpy as np

data = np.random.lognormal(mean=0, sigma=1, size=10)

print(data)
```
