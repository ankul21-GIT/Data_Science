# Power Law Distribution




A Power Law Distribution is a distribution where small values occur very frequently and very large values occur rarely, but the large values are extremely significant.

In simple words:

👉 Many small events + few very large events

Example pattern:

80% small values
20% very large values

This is often called the Pareto principle (80–20 rule).

Mathematical Formula

P(X=x) \propto x^{-\alpha}

Where:

x = value

α (alpha) = power law exponent

∝ means proportional

This means probability decreases as x increases.

Shape of Power Law Distribution

Characteristics:

Heavy tail

Many small values

Few extremely large values

Often appears as a straight line in log-log plots

```python
import numpy as np

data = np.random.power(5,1000)

print(data[:10])
```

**Generate Power Law Data**

```python
import numpy as np

# generate power law distributed data
alpha = 3   # shape parameter
size = 1000

data = np.random.power(alpha, size)

print(data[:10])   # first 10 values
```

**Plot the Distribution**

```python
import numpy as np
import matplotlib.pyplot as plt

alpha = 3
data = np.random.power(alpha, 10000)

plt.hist(data, bins=50)
plt.title("Power Law Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

**Log-Log Plot**

```python
import numpy as np
import matplotlib.pyplot as plt

alpha = 2.5
data = np.random.power(alpha, 10000)

hist, bins = np.histogram(data, bins=50)

plt.loglog(bins[:-1], hist)
plt.title("Power Law Log-Log Plot")
plt.xlabel("log(Value)")
plt.ylabel("log(Frequency)")
plt.show()
```

**Real Example: Pareto Distribution (Power Law)**

```python
import numpy as np

# Pareto distribution
alpha = 2
data = np.random.pareto(alpha, 1000)

print(data[:10])
```
