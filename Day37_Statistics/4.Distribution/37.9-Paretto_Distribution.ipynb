# Pareto Distribution

The Pareto Distribution models situations where a small number of causes produce most of the effects.



This idea is called the Pareto Principle (80–20 rule).

Examples:

- 20% customers generate 80% revenue

- 20% bugs cause 80% software failures

- 20% products produce 80% sales

Pareto Distribution Formula

f(x)=\frac{\alpha x_m^{\alpha}}{x^{\alpha+1}}, \quad x \ge x_m

Where:
𝑥 = minimum possible value

α = shape parameter (Pareto index)

x = random variable

**Mean of Pareto Distribution**

E(X)=\frac{\alpha x_m}{\alpha-1}, \quad \alpha>1

**Variance**

Var(X)=\frac{\alpha x_m^2}{(\alpha-1)^2(\alpha-2)}, \quad \alpha>2

```python
import numpy as np

alpha = 2.5

data = np.random.pareto(alpha, 1000)

print(data[:10])
```

```python
# Visualizing Pareto Distribution
import numpy as np
import matplotlib.pyplot as plt

alpha = 2.5
data = np.random.pareto(alpha, 10000)

plt.hist(data, bins=50)
plt.title("Pareto Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
