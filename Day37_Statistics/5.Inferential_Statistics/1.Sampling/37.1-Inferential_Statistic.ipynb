# Inferential Statistics

Inferential statistics is a branch of statistics used to make conclusions or predictions about a population based on a sample of data.

Instead of collecting data from an entire population (which is expensive or impossible), we analyze a sample and infer results for the whole population.

📌 Example

Suppose:

A city has 1,000,000 people (population)

You survey 1,000 people (sample)

From this sample you estimate:

average income

voting preference

product preference

This process is inferential statistics.

```python
import numpy as np

# population data
population = np.random.normal(50,10,10000)

# sample from population
sample = np.random.choice(population,100)

print("Population mean:", np.mean(population))
print("Sample mean:", np.mean(sample))
```
