# Simple Random Sampling

Every member of the population has an equal probability of being selected.

Example:

If a population has 1000 people, each person has 1/1000 chance of selection.

Example uses:

Lottery system

Random surveys

```python
import numpy as np

population = np.arange(1,1001)

sample = np.random.choice(population,50,replace=False)

print(sample)
```
