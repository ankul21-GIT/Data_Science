# Stratified Sampling

Population is divided into subgroups (strata) based on characteristics.

Example:

Students grouped by:

Male

Female

Then samples are taken from each group proportionally.

This ensures better representation.

```python
import pandas as pd

data = {
"Gender":["M","F","M","F","M","F","M","F"],
"Marks":[70,80,65,90,75,85,60,88]
}

df = pd.DataFrame(data)

sample = df.groupby("Gender").sample(2)

print(sample)
```
