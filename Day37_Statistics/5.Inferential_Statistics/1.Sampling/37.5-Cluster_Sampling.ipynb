# Cluster Sampling

Population is divided into clusters, then entire clusters are selected.

Example:

City divided into:

Area A

Area B

Area C

Area D

Instead of selecting individuals, we randomly choose some clusters.

```python
import pandas as pd
import numpy as np

# create dataset
np.random.seed(42)

data = pd.DataFrame({
    "student_id": range(1,101),
    "cluster": np.repeat(range(1,11),10),  # 10 clusters
    "score": np.random.randint(50,100,100)
})

print(data.head())
```
