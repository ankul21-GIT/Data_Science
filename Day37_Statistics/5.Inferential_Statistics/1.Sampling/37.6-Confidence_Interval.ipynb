# Confidence Interval (CI)

A confidence interval is a range of values that is likely to contain the true population parameter (such as the population mean) with a certain level of confidence.

Instead of giving a single estimate, we give a range where the true value probably lies.

Example

Suppose you survey 100 students and find:

Sample mean height = 170 cm

Using statistics you calculate a 95% confidence interval:

168 cm to 172 cm

Meaning:
We are 95% confident that the true average height of the whole population lies between 168 and 172 cm.

Important:
It does not mean there is a 95% probability the mean is in that interval. It means the method produces correct intervals 95% of the time.

📐 Confidence Interval Formula (Mean)

\bar{x} \pm z\left(\frac{\sigma}{\sqrt{n}}\right)

Where:

𝑥 = sample mean

z = z-score based on confidence level

σ = population standard deviation

n = sample size

```python
import numpy as np
import scipy.stats as stats

# sample data
data = [165,170,172,168,174,169,171,167,173,170]

mean = np.mean(data)
std = np.std(data, ddof=1)
n = len(data)

confidence = 0.95

interval = stats.t.interval(confidence, n-1, loc=mean, scale=std/np.sqrt(n))

print("Mean:", mean)
print("Confidence Interval:", interval)
```

**Why Confidence Intervals Matter**
