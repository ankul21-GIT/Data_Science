# Central Limit Theorem (CLT)

The Central Limit Theorem is one of the most important results in Inferential Statistics.

It explains how we can use a sample to make conclusions about a population.

Definition

The Central Limit Theorem states:

If we take many random samples from a population, the sampling distribution of the sample mean will approach a normal distribution, even if the original population is not normally distributed.



Key Idea of CLT

Even if the original population distribution is:

skewed

uniform

exponential

The distribution of sample means becomes normal when the sample size is large.

Usually:

n ≥ 30
Example

Suppose a population of incomes is skewed.

If we take many samples of size 30 or more, the mean of those samples will follow a normal distribution.

This is why normal distribution appears everywhere in statistics.

**Why CLT Is Important**

CLT allows us to:

- estimate population parameters

- build confidence intervals

- perform hypothesis testing

- use Z-tests and t-tests

This is the foundation of inferential statistics.

Simple Example

Population (not normal):

2, 4, 6, 20, 30, 100

Take many samples and compute their means:

Sample means → 12, 14, 13, 15, 14 ...

These sample means will form a normal distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

population = np.random.exponential(scale=2, size=10000)

sample_means = []

for i in range(1000):
    sample = np.random.choice(population, 30)
    sample_means.append(np.mean(sample))

plt.hist(sample_means, bins=30)
plt.title("Central Limit Theorem")
plt.show()
```
