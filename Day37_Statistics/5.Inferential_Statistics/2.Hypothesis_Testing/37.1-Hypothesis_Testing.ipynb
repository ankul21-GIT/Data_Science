# Hypothesis Testing

Hypothesis testing is a statistical method used to test an assumption about a population parameter (like mean, proportion, etc.).

In simple words:

👉 We use sample data to check whether a claim about the population is true or false.

Example

Suppose a company claims:

Average battery life of a phone = 10 hours

You test 50 phones and calculate the sample mean.

Using hypothesis testing, you check:

👉 Is the company claim true or not?

*Two Types of Hypotheses*

**Null Hypothesis (H₀)**

This is the default assumption.

It usually states no effect or no difference.

Example:

H₀: μ = 10

Meaning the average battery life is 10 hours.

**Alternative Hypothesis (H₁ or Ha)**

This is what we want to test.

Example:

H₁: μ ≠ 10

Meaning the battery life is not 10 hours.

```python
from scipy import stats
import numpy as np

data = [102,105,98,110,100,107,103,104]

t_stat, p_value = stats.ttest_1samp(data, 100)

print("p-value:", p_value)
```
