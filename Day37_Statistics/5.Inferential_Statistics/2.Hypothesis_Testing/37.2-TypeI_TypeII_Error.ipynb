# Type I Error (False Positive)

**A Type I error occurs when we:**

👉 Reject the null hypothesis even though it is actually true.

In simple words:

We think something changed, but actually nothing changed.

Example:

A medical test says a person has a disease, but the person is actually healthy.

So:

H₀ is true

but we rejected it

The probability of Type I error is called the significance level (α).

Usually:

α = 0.05

Meaning 5% chance of Type I error.

```python
import numpy as np
from scipy import stats

np.random.seed(0)

true_mean = 50
sample_size = 30
alpha = 0.05
experiments = 1000

type1_errors = 0

for _ in range(experiments):
    
    # generate sample where H0 is actually true
    sample = np.random.normal(true_mean, 10, sample_size)
    
    # perform t-test
    t_stat, p_value = stats.ttest_1samp(sample, 50)
    
    if p_value < alpha:
        type1_errors += 1

print("Type I Error Rate:", type1_errors/experiments)
```

# Type II Error (False Negative)

A Type II error occurs when we:

👉 Fail to reject the null hypothesis even though it is false.

In simple words:

We think nothing changed, but actually something changed.

Example:

A medical test says a person is healthy, but the person actually has the disease.

So:

H₀ is false

but we did not reject it

The probability of Type II error is called:

β (beta)

**Simple Real-Life Example**

Suppose a company tests a new medicine.

**Null hypothesis:**

H₀: Medicine does not work

Alternative hypothesis:

H₁: Medicine works

Now possible errors:

**Type I Error:**

We say medicine works

but actually it does not.

**Type II Error:**

We say medicine does not work

but actually it works.

```python
import numpy as np
from scipy import stats

np.random.seed(0)

true_mean = 55   # real mean different from H0
sample_size = 30
alpha = 0.05
experiments = 1000

type2_errors = 0

for _ in range(experiments):

    sample = np.random.normal(true_mean, 10, sample_size)

    t_stat, p_value = stats.ttest_1samp(sample, 50)

    if p_value > alpha:
        type2_errors += 1

print("Type II Error Rate:", type2_errors/experiments)
```
