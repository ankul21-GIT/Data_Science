# Z-test

A Z-test is a statistical test used to determine whether there is a significant difference between a sample mean and a population mean, when the population standard deviation is known and the sample size is large (usually n ≥ 30).





***Real-World Use***

Z-tests are used in:

- Product quality testing
- Market research
- A/B testing
- Medical studies



```python
import numpy as np
from statsmodels.stats.weightstats import ztest

data = [52,55,50,48,54,53,51,56,52,49]

z_stat, p_value = ztest(data, value=50)

print("Z statistic:", z_stat)
print("P-value:", p_value)
```
