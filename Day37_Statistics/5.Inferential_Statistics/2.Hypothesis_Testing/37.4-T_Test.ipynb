# T-Test

A T-test is a statistical test used to determine whether there is a significant difference between the means of two groups, when the population standard deviation is unknown and the sample size is usually small (n < 30).

📐 T-Test Formula

t = \frac{\bar{x} - \mu}{s / \sqrt{n}}

Where:

𝑥 = sample mean

μ = population mean

s = sample standard deviation

n = sample size

📌 Example

Suppose:

Claimed average exam score = 70

Sample mean = 75

Sample standard deviation = 10

Sample size = 25

Then we calculate the t-value using the formula and compare it with the t-distribution table.



Example of Each

*One-Sample T-Test*

Check if the average height of students differs from 170 cm.

*Independent T-Test*

Compare average marks of two different classes.

*Paired T-Test*

Compare before-treatment vs after-treatment results.



```python
from scipy import stats

group1 = [70,72,68,65,74,71]
group2 = [80,85,78,82,79,81]

t_stat, p_value = stats.ttest_ind(group1, group2)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```
