# Chi-Square Test


A Chi-Square test (χ² test) is a statistical test used to determine whether there is a significant relationship between categorical variables.

It compares the observed frequencies with the expected frequencies.

📐 Chi-Square Formula

\chi^2 = \sum \frac{(O - E)^2}{E}

Where:

𝑂
O = Observed frequency

𝐸
E = Expected frequency

The test measures how much the observed data differs from what we expect.



```python
import scipy.stats as stats
import numpy as np

# observed frequencies
data = np.array([[30,20],
                 [10,40]])

chi2, p, dof, expected = stats.chi2_contingency(data)

print("Chi-square value:", chi2)
print("P-value:", p)
```

# Real-World Applications

Chi-Square tests are used in:

- Market research

- Survey analysis

- Medical research

- Customer behavior analysis

Example: companies like Amazon analyze whether customer gender affects product preference.

**Important Condition**

Chi-Square test works when:

- Data is categorical

- Observations are independent

- Expected frequency is ≥ 5
