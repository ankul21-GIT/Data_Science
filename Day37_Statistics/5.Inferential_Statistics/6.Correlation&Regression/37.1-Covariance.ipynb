# Covariance

Covariance measures how two variables change together.

It tells us whether the variables increase together or move in opposite directions.

📌 Example

Suppose we study:

Study hours

Exam scores

If students who study more get higher scores, then covariance is positive.



Real-World Use

Covariance is used in:

- Finance (stock relationships)

- Machine learning

- Feature relationships

- Portfolio risk analysis

Example: analysts study covariance between stocks of companies like Apple and Microsoft.

```python
import numpy as np

study_hours = [2,4,6,8]
marks = [50,60,65,80]

covariance = np.cov(study_hours, marks)

print(covariance)
```
