# Correlation

Correlation measures the strength and direction of the relationship between two variables.

Unlike covariance, correlation gives a standardized value between -1 and +1.

📐 Correlation Formula (Pearson Correlation)

r = \frac{\text{Cov}(X,Y)}{\sigma_X \sigma_Y}

Where:

r = correlation coefficient

Cov(X,Y) = covariance between variables

X = standard deviation of X

Y = standard deviation of Y

| Correlation Value | Meaning                      |
| ----------------- | ---------------------------- |
| +1                | Perfect positive correlation |
| 0                 | No correlation               |
| -1                | Perfect negative correlation |


Example

Suppose we study:

Study Hours	Exam Marks
2	50
4	60
6	70
8	80

As study hours increase, marks also increase.

This shows positive correlation.

📊 Types of Correlation

- Type	                -  Meaning
- Positive Correlation	- Both variables increase together
- Negative Correlation	- One increases while the other decreases
- Zero Correlation	   -  No relationship

Example:

- Positive → study hours vs marks
- Negative → exercise vs body fat


```python
import numpy as np

study_hours = [2,4,6,8]
marks = [50,60,70,80]

correlation = np.corrcoef(study_hours, marks)

print(correlation)
```

```python
# Correlation Heatmap
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

data = {
"StudyHours":[2,4,6,8],
"Marks":[50,60,70,80]
}

df = pd.DataFrame(data)

sns.heatmap(df.corr(), annot=True)

plt.show()
```
