```python
ratings = final_df.groupby(['Aggregate rating','Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
ratings
```

## Observation
1. When Rating is between 4.5 to 4.9 ---> Excellent
2. When Rating are between 4.0 to 3.4 ---> very good
3. when Rating is between 3.5 to 3.9 ---> good
4. when Rating is between 3.0 to 3.4 ---> average
5. when Rating is between 2.5 to 2.9 ---> average
6. when Rating is between 2.0 to 2.4 ---> Poor


```python
ratings.head()
```

```python
sns.barplot(x='Aggregate rating',y='Rating Count',data=ratings)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (12,6)
sns.barplot(x='Aggregate rating',y='Rating Count',data=ratings,palette="BuGn")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (12,6)
sns.barplot(x='Aggregate rating',y='Rating Count',data=ratings,palette="icefire", hue='Rating color')
plt.show()
```

```python
sns.barplot(x='Aggregate rating',y='Rating Count',data=ratings,hue='Rating color',palette=['white','red','orange','yellow','green'])
```
